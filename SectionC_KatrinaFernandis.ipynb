17. Swiggy Order Calculator. Take item price and quantity as input. Compute total = price ×
quantity. Add 18% GST. Add ₹40 delivery charge if total (before delivery) is below ₹500, else delivery
is FREE. Print all components separately and final amount in this format:
Subtotal: ₹450.00
GST (18%): ₹81.00
Delivery: ₹40.00
------------------------
Total: ₹571.00


```python
price = float(input("Enter item price:"))
quantity = int(input("Enter quantity:"))

#calculate subtotal
subtotal = price * quantity

#calculate GST
gst = subtotal * 0.18

#calculate delivery charge
if subtotal < 500:
    delivery=40
else:
    delivery=0

#calculate total
total=subtotal+gst+delivery

print(f"\nSubtotal: {subtotal:.2f}")
print(f"GST (18%):{gst:.2f}")
print(f"Delivery:{delivery:.2f}")
print("-------------------------")
print(f"Toatal:{total:.2f}")
```

Q19. Take a 4-digit number as input. Print the sum of its digits – WITHOUT converting it to a string.
(Use only integer arithmetic: //, %.)


```python
number=int(input("Enter a 4-digit number:"))
sum_digits=0
while number >0:
  sum_digits+=number % 10
  number=number//10
print(f"The sun of digits is:{sum_digits}")
```

Q21. Indian Income Tax Calculator (New Regime, FY 2024–25). Take annual income as input.
Compute tax using these slabs:
Up to ₹3,00,000 : Nil
₹3,00,001 – ₹7,00,000 : 5%
₹7,00,001 – ₹10,00,000 : 10%
₹10,00,001 – ₹12,00,000 : 15%
₹12,00,001 – ₹15,00,000 : 20%
Above ₹15,00,000 : 30%

```python
i=int(input("Enter income(₹):"))
t=0
b=[]
p=0
for s, r in [(300000,0),(700000,0.05),(1000000,0.1),(1200000,0.15),(1500000,0.2),(float('inf'),0.3)]:
  if i <= p: break
  a=min(i-p,s-p)
  b.append((p,s,a,a*r))
  t+=a*r
  p=s
print("Breakdown:")
for l, u, a, st in b:
  print(f"₹{l+1:,}-₹{u:,}:{int(r*100)}% → ₹{st:,.2f}")
print(f"\nTotal Tax:₹{t:,.2f}")
```

Q24. Cricket Strike Rate Classifier. Take runs scored and balls faced. Compute strike rate =
(runs/balls) × 100. Classify:
SR > 150 → Explosive (T20 finisher)
100 < SR ≤ 150 → Aggressive (Top order T20)
75 ≤ SR ≤ 100 → Steady (ODI anchor)
SR < 75 → Defensive (Test match)

```python
runs = int(input("Enter runs scored:"))
balls = int(input("Enter balls faced:"))
if balls == 0:
  print("Balls faced cannot be zero.")
else:
  strike_rate=(runs/balls)*100
  if strike_rate > 150:
    classification ="Explosive (T20 finisher)"
  elif 100 < strike_rate<=150:
    classification ="Aggressive(Top order T20)"
  elif 75 <= strike_rate<= 100:
    classification = "Steady(ODI anchor)"
  else:
    classification = "Defensive (Test match)"
print(f"Strike Rate:{strike_rate:.2f}")
print(f"Classification:{classification}")
```

Q27. Print this number diamond for N = 5 (the digits go up to N in the middle row):

```python
N=5
#upper half
for i in range(1,N+1):
  for _ in range(N-i):
    print(" ",end=" ")
  for j in range(1,i+1):
    print(j,end=" ")
  print()
#lower half
for i in range(N-1,0,-1):
  for _ in range(N-i):
    print(" ",end=" ")
  for j in range(1,i+1):
    print(j,end=" ")
  print()


```

Q28. Print a hollow square of size N (only the border is *, inside is blank). For N = 5:

```python
N=5
for i in range(N):
  for j in range(N):
    if i == 0 or i == N - 1 or j == 0 or j == N - 1:
      print("*",end="")
    else:
      print(" ",end="")
  print()
```
